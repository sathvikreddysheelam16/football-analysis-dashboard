```python
# Create a copy for cleaning

df_clean = df.copy()

# Remove unnecessary URL columns

remove_cols = [
    'player_url',
    'player_face_url',
    'club_logo_url',
    'club_flag_url',
    'nation_logo_url',
    'nation_flag_url'
]

df_clean = df_clean.drop(columns=remove_cols)

df_clean.head()
```

```python
df_clean.shape
```

```python
df_clean.isnull().sum().sort_values(ascending=False).head(15)
```

```python
df_clean.shape
```

```python
df_clean.isnull().sum().sort_values(ascending=False).head(20)
```

```python
# Remove columns with more than 50% missing values

missing_percentage = df_clean.isnull().mean()

columns_to_drop = missing_percentage[missing_percentage > 0.5].index

df_clean = df_clean.drop(columns=columns_to_drop)

df_clean.shape
```

```python
# Fill numerical missing values with median

numeric_cols = df_clean.select_dtypes(include=['int64','float64']).columns

df_clean[numeric_cols] = df_clean[numeric_cols].fillna(
    df_clean[numeric_cols].median()
)


# Fill text missing values

text_cols = df_clean.select_dtypes(include=['object']).columns

df_clean[text_cols] = df_clean[text_cols].fillna("Unknown")
```
